```python
import matplotlib.pyplot as plt
```

```python
plt.plot([1,2,3,4])
plt.show()
```

```python
plt.plot([1,2,3,4], [4,5,6,7])
plt.show()
```

```python
x = [1,2,3,4]
sql_marks = [90, 92, 96, 98]
python_marks = [93, 95, 98, 100]

plt.plot(x, sql_marks, label = "sql marks")
plt.plot(x, python_marks, label = "python marks")

plt.title("marks corresponding to no of hrs studied")

plt.xlabel("no of hrs studied")
plt.ylabel("marks")

plt.xticks([1,2,3,4])
plt.yticks([90, 95, 100])

plt.legend()

plt.show()
```
