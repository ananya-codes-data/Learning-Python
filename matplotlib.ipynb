## Multiple plots in same figure

```python
x = [1,2,3,4]
sql_marks = [90, 92, 96, 98]
python_marks = [93, 95, 98, 100]

plt.plot(x, sql_marks, 'c*--' , label = "sql marks")
plt.plot(x, python_marks, 'y^:' , label = "python marks")

plt.title("marks corresponding to no of hrs studied",
          fontdict={'fontname' : 'comic sans ms', 'fontweight': 'bold', 'fontsize' : 15, 'color' : 'blue'})

plt.xlabel("no of hrs studied",
           fontdict={'fontname' : 'times new roman', 'fontweight': 'bold', 'fontsize' : 12, 'color' : 'red'})
plt.ylabel("marks",
           fontdict={'fontname' : 'times new roman', 'fontweight': 'bold', 'fontsize' : 12, 'color' : 'red'})

plt.xticks([1,2,3,4])
plt.yticks([90, 95, 100])

plt.legend(loc = "lower right")

plt.savefig("marks.png")

plt.show()
```

```python
x = [1,2,3,4]
sql_marks = [90, 92, 96, 98]
python_marks = [93, 95, 98, 100]

plt.plot(x, sql_marks, label = "sql marks", color = "cyan", linewidth = 2, marker = "*", markersize = 10)
plt.plot(x, python_marks, label = "python marks", color = "yellow", linewidth = 1, marker = "^", linestyle = "--")

plt.title("marks corresponding to no of hrs studied",
          fontfamily ="comic sans ms")

plt.xlabel("no of hrs studied")
plt.ylabel("marks")

plt.xticks([1,2,3,4])
plt.yticks([90, 95, 100])

plt.legend(loc = "lower right")

plt.show()
```

```python
# evenly sampled time at 200ms intervals
t = np.arange(1, 6, 1)

# red dashes, blue squares and green triangles
plt.plot(t, t, 'r--', t, t**2, 'bs', t, t**3, 'g^')
plt.show()
```

##  Plots in different figures

```python
n = np.arange(1, 6, 1)

plt.figure()
plt.plot(n, n, 'r--')
plt.show()

plt.figure()
plt.plot(n, n**2, 'bs')
plt.show()

plt.figure()
plt.plot(n, n**3, 'g^')
plt.show()
```

## Subplots in a figure

```python
m = np.arange(1, 6, 1)

plt.figure()

plt.subplot(1, 3, 1)
plt.plot(m, m, 'r--')
    
plt.subplot(1, 3, 2)
plt.plot(m, m**2, 'bs')

plt.subplot(1, 3, 3)
plt.plot(m, m**3, 'g^')

plt.tight_layout()

plt.show()
```

```python
q = np.arange(1, 6, 1)

plt.figure()

plt.subplot(3, 1, 1)
plt.plot(q, q, 'r--')
plt.xlabel("q values")
plt.ylabel("output values")
plt.title("linear subplot")
    
plt.subplot(3, 1, 2)
plt.plot(q, q**2, 'bs')
plt.xlabel("q values")
plt.ylabel("output values")
plt.title("quadratic subplot")

plt.subplot(3, 1, 3)
plt.plot(q, q**3, 'g^')
plt.xlabel("q values")
plt.ylabel("output values")
plt.title("cubic subplot")

plt.suptitle("different functions of q")
plt.tight_layout()

plt.show()
```

```python
b = np.arange(1, 6, 1)

plt.figure()

plt.subplot(1, 3, 1)
plt.plot(b, b, 'r--')
plt.xlabel("b values")
plt.ylabel("output values")
plt.title("linear subplot")
    
plt.subplot(1, 3, 2)
plt.plot(b, b**2, 'bs')
plt.xlabel("b values")
plt.ylabel("output values")
plt.title("quadratic subplot")

plt.subplot(1, 3, 3)
plt.plot(b, b**3, 'g^')
plt.xlabel("b values")
plt.ylabel("output values")
plt.title("cubic subplot")

plt.suptitle("different functions of b")
plt.tight_layout()

plt.show()
```

## Axes, Generation of Subplots

```python
# axes - x axis, y axis, labels, ticks

x = np.linspace(1, 11, 100)

fig, axes = plt.subplots(2, 3)

axes[0, 0].plot(x, np.sin(x))
axes[0, 0].set_title("Sin func")
axes[0, 0].set_xlabel("x")
axes[0, 0].set_ylabel("y")

axes[0, 1].plot(x, np.cos(x))
axes[0, 1].set_title("Cos func")
axes[0, 1].set_xlabel("x")
axes[0, 1].set_ylabel("y")

axes[0, 2].plot(x, np.tan(x))
axes[0, 2].set_title("Tan func")
axes[0, 2].set_xlabel("x")
axes[0, 2].set_ylabel("y")

axes[1, 0].plot(x, np.exp(x))
axes[1, 0].set_title("Exp func")
axes[1, 0].set_xlabel("x")
axes[1, 0].set_ylabel("y")

axes[1, 1].plot(x, np.sqrt(x))
axes[1, 1].set_title("Sqrt func")
axes[1, 1].set_xlabel("x")
axes[1, 1].set_ylabel("y")

axes[1, 2].plot(x, np.log(x))
axes[1, 2].set_title("Log func")
axes[1, 2].set_xlabel("x")
axes[1, 2].set_ylabel("y")

plt.tight_layout()

plt.show()
```

## Scatter Plots

```python
x = np.random.rand(50) * 10
y = np.random.rand(50) * 10

plt.scatter(x,y, edgecolors='black', linewidths=2, color='orange', s=100, alpha=0.7)

plt.legend(["random"], loc = "upper right")

plt.grid(True)

plt.show()
```

## Bar Plots

```python
courses = ['Excel', 'Python', 'SQL', 'Power BI', 'ML', 'GenAI']
marks = [53, 80, 95, 65, 72, 45]

plt.bar(courses, marks, width = 0.5, align = 'center', color = 'red', edgecolor = 'black', linewidth = 2, hatch='*')

plt.yticks([20, 40, 60, 80, 100])

plt.show()
```

```python
courses = ['Excel', 'Python', 'SQL', 'Power BI', 'ML', 'GenAI']
marks = [53, 80, 95, 65, 72, 45]
hatches = ['/', '*', 'o', '\\', '|', '-']

bars = plt.bar(courses, marks, width = 0.5, align = 'center', color = 'red', edgecolor = 'black', linewidth = 2, hatch='*')

plt.yticks([20, 40, 60, 80, 100])

for bar, hatch in zip(bars, hatches):
    bar.set_hatch(hatch)

plt.show()
```
