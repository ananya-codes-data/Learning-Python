```python
import pandas as pd
```

```python
df = pd.read_excel("https://raw.githubusercontent.com/ananya-codes-data/Learning-Python/main/Guided_projects/PAN_card_cleaning_%26_validation_project/PAN%20Number%20Validation%20Dataset.xlsx")
df
```

```python
print(df.head(10))
```

```python
print('Total records = ',len(df))
```

```python
df['Pan_Numbers'] = df['Pan_Numbers'].astype('string').str.strip().str.upper()
```

```python
print(df.head(10))
```

```python
# finding number of empty strings
print((df['Pan_Numbers'] == '').sum())
```

```python
# finding total number of null values
print(df.isnull().sum())
```

```python
df = df.replace({'Pan_Numbers': ''}, pd.NA).dropna(subset = 'Pan_Numbers')
```

```python
print('Total records = ',len(df))
```

```python
print('Unique records = ',df['Pan_Numbers'].nunique())
```

```python
df = df.drop_duplicates(subset='Pan_Numbers', keep='first')
print('Total records = ',len(df))
```
