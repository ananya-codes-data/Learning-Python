```python
import pandas as pd
import numpy as np
```

```python
# Load the messy data

df = pd.read_csv('https://raw.githubusercontent.com/ananya-codes-data/Learning-Python/refs/heads/main/Guided_projects/Marketing_campaign_data_cleaning/marketing_campaign_data_messy.csv')

print(f"Loaded Dataset : {df.shape[0]} rows, {df.shape[1]} columns")
```

```python
# STEP 1: HEADER CLEANING

print(df.columns.tolist())

df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')

print("FIX APPLIED")
print(df.columns.tolist())
```

```python
# STEP 2: TYPE CONVERSION & CURRENCY CLEANING

dirty_spend_mask = df['spend'].astype(str).str.contains(r'\$')
print(df.loc[dirty_spend_mask, ['campaign_id', 'spend']])

df['spend'] = df['spend'].astype(str).str.replace(r'[^\d.-]', '', regex= True)
df['spend'] = pd.to_numeric(df['spend'])

print('FIX APPLIED')
print(df.loc[dirty_spend_mask, ['campaign_id', 'spend']])
```

```python
# STEP 3: CATEGORICAL TYPOS (FUZZY LOGIC)

print(df['channel'].unique())

cleanup_map = {
    'Facebok': 'Facebook',
    'Tik_Tok': 'TikTok',
    'E-mail': 'Email',
    'Insta_gram': 'Instagram',
    'Gogle': 'Google Ads',
    'N/A': np.nan   # Handling the ghost value here too
}

df['channel'] = df['channel'].replace(cleanup_map)

print("FIX APPLIED")
print(df['channel'].unique())
```

```python
# STEP 4: HANDLING MIXED BOOLEANS

print(df['active'].unique())

bool_map = {
    'Yes': True,
    'Y': True,
    '1': True,
    1: True,
    'No': False,
    '0': False,
    0: False
}

df['active'] = df['active'].map(bool_map).fillna(False).astype(bool)

print("FIX APPLIED")
print(df['active'].unique())
```

```python
# STEP 5: DATE PARSING

print(df['start_date'].dtype)
print(df['end_date'].dtype)

df['start_date'] = pd.to_datetime(df['start_date'], errors= 'coerce')
df['end_date'] = pd.to_datetime(df['end_date'], dayfirst= True, errors= 'coerce')

print("FIX APPLIED")

print(df['start_date'].dtype)
print(df['end_date'].dtype)
```

```python
df = df.loc[:, ~df.columns.duplicated()]
df
```

```python
# STEP 6: LOGICAL INTEGRITY (CLICKS vs IMPRESSIONS)

impossible_mask = df['clicks'] > df['impressions']
print(df.loc[impossible_mask, ['campaign_id', 'impressions', 'clicks']])
```

```python
# STEP 7: LOGICAL INTEGRITY (TIME TRAVEL)

time_travel_mask = df['start_date'] > df['end_date']
print(df.loc[time_travel_mask, ['campaign_id', 'start_date', 'end_date']].head(5))

# assuming that campaign last for 30 days

df.loc[time_travel_mask, 'end_date'] = df.loc[time_travel_mask, 'start_date'] + pd.Timedelta(days = 30)

print("FIX APPLIED")

print(df.loc[time_travel_mask, ['campaign_id', 'start_date', 'end_date']].head(5))
```

```python
# STEP 8: HANDLING OUTLIERS (WINSORIZING) #### error -- need to fix this code

Q1 = df['spend'].quantile(0.25)
Q3 = df['spend'].quantile(0.75)

IQR = Q3 - Q1

upper_limit = Q3 + (3 * IQR)
print("upper limit =", upper_limit)
upper_outlier_mask = df['spend'] > upper_limit
print(df.loc[upper_outlier_mask, ['campaign_id', 'spend']].head(2))

print("FIX APPLIED")
df.loc[upper_outlier_mask, 'spend'] = upper_limit
print(df.loc[upper_outlier_mask, ['campaign_id', 'spend']].head(2))


lower_limit = Q1 - (3 * IQR)
print("lower limit =", lower_limit)
lower_outlier_mask = df['spend'] > lower_limit
print(df.loc[lower_outlier_mask, ['campaign_id', 'spend']].head(2))

print("FIX APPLIED")
df.loc[lower_outlier_mask, 'spend'] = lower_limit
print(df.loc[lower_outlier_mask, ['campaign_id', 'spend']].head(2))
```
