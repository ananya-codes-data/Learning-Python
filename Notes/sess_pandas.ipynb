# Pandas Series (S-16)

Importing libraries

```python
import pandas as pd
import numpy as np
```

Series from lists

```python
# string
country = ['India', 'Pakistan', 'Nepal', 'USA', 'Srilanka']

pd.Series(country)
```

```python
# integers
runs = [28, 13, 78, 65, 84, 92]

run_series = pd.Series(runs)
run_series
```

```python
# custom index
marks = [67, 57, 89, 100]

subjects = ['english', 'science', 'maths', 'hindi']

# marks as value subject as index

pd.Series(marks, index= subjects)
```

```python
marks = pd.Series(marks, index= subjects, name= 'Subject wise marks')
marks
```

Series from dict

```python
marks = {
    'english': 67,
    'science': 57,
    'maths': 89,
    'hindi': 100
}

marks_series = pd.Series(marks, name= 'Subject wise marks')
marks_series
```
