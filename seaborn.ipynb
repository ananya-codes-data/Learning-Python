## Import libraries

```python
import seaborn as sns
import pandas as pd
import numpy as np
```

## dataset names from seaborn

```python
sns.get_dataset_names()
```

## using penguins dataset from seaborn

```python
penguins = sns.load_dataset('penguins')
penguins.head()
```

## categorical variables

```python
penguins["species"].value_counts()
```

```python
penguins["island"].value_counts()
```

## Scatter Plot

```python
sns.scatterplot(data= penguins, x= "flipper_length_mm", y= "body_mass_g", hue= "island")
```

```python
sns.scatterplot(data= penguins, x= "flipper_length_mm", y= "body_mass_g", hue= "sex")
```

### set_style

- darkgrid
- dark
- whitegrid
- white
- ticks

```python
sns.set_style("darkgrid")

sns.scatterplot(data= penguins, x= "flipper_length_mm", y= "body_mass_g", hue= "island")
```

### despine()

```python
sns.set_style("ticks")

sns.scatterplot(data= penguins, x= "flipper_length_mm", y= "body_mass_g", hue= "island")

sns.despine(left= True, bottom= True)
```

```python
sns.set_style("ticks")

sns.scatterplot(data= penguins, x= "flipper_length_mm", y= "body_mass_g", hue= "island")

sns.despine()
```

### set_context

- notebook
- paper
- talk
- poster

```python
sns.set_style("ticks")

sns.set_context('notebook')

sns.scatterplot(data= penguins, x= "flipper_length_mm", y= "body_mass_g", hue= "island")

sns.despine()
```
